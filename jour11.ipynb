```python
import pandas as p

with open("etudiant1.csv") as f:
    a=p.read_csv(f)
    b=a[a["filiere"]=="Ing1"]
    c=a["filiere"].value_counts()
    d=a.groupby(["filiere","nom"])["age"].mean()
    e=a.dropna()
    r=p.options.display.max_rows
    s=a["poids"].mean()
    ri=a.isna().sum()
    my=a.fillna({"age":19,"poids":s},inplace=True)
    pi=p.to_datetime(a['date'], format='mixed')
print(b)
print(c)
print(d)
print(e)
print(a)
print(r)
print(my)
print(ri)
print(pi)
```
